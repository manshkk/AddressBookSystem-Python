```python
from models.contact import Contact
from models.address_book import AddressBook

from services.file_service import FileService
from services.csv_service import CSVService
from services.json_service import JSONService

from exceptions.custom_exceptions import (
    DuplicateContactException,
    ContactNotFoundException,
    InvalidPhoneNumberException,
    InvalidEmailException,
    InvalidZipCodeException,
)
```

```python
def add_contact():

    try:

        contact = Contact(
            input("First Name : "),
            input("Last Name : "),
            input("Address : "),
            input("City : "),
            input("State : "),
            input("Zip Code : "),
            input("Phone : "),
            input("Email : "),
        )

        book.add_contact(contact)

        print("\nContact Added Successfully.")

    except DuplicateContactException as error:
        print(error)

    except InvalidPhoneNumberException as error:
        print(error)

    except InvalidEmailException as error:
        print(error)

    except InvalidZipCodeException as error:
        print(error)

    except Exception as error:
        print(error)
```

```python
def edit_contact():

    try:

        first_name = input("Enter First Name of Contact to Edit : ")
        last_name = input("Enter Last Name of Contact to Edit : ")

        contact = Contact(
            input("New First Name : "),
            input("New Last Name : "),
            input("New Address : "),
            input("New City : "),
            input("New State : "),
            input("New Zip Code : "),
            input("New Phone : "),
            input("New Email : ")
        )

        book.edit_contact(first_name, last_name, contact)

        print("\nContact Updated Successfully.")

    except ContactNotFoundException as error:
        print(error)

    except InvalidPhoneNumberException as error:
        print(error)

    except InvalidEmailException as error:
        print(error)

    except InvalidZipCodeException as error:
        print(error)

    except Exception as error:
        print(error)
```

```python
def delete_contact():

    try:

        first_name = input("Enter First Name : ")
        last_name = input("Enter Last Name : ")

        book.delete_contact(first_name, last_name)

        print("\nContact Deleted Successfully.")

    except ContactNotFoundException as error:
        print(error)

    except Exception as error:
        print(error)
```

```python
def display_contacts():

    contacts = book.display_contacts()

    if not contacts:
        print("\nNo Contacts Found.")
        return

    print("\nContact List\n")

    for contact in contacts:
        print(contact)
        print("-" * 50)
```

```python
def search_by_city():

    city = input("Enter City : ")

    contacts = book.search_by_city(city)

    if not contacts:
        print("\nNo Contact Found.")
        return

    print()

    for contact in contacts:
        print(contact)
        print("-" * 50)
```

```python
def search_by_state():

    state = input("Enter State : ")

    contacts = book.search_by_state(state)

    if not contacts:
        print("\nNo Contact Found.")
        return

    print()

    for contact in contacts:
        print(contact)
        print("-" * 50)
```

```python
def view_by_city():

    city_contacts = book.view_by_city()

    if not city_contacts:
        print("\nNo Contacts Found.")
        return

    print("\nContacts Grouped By City\n")

    for city, contacts in city_contacts.items():

        print(f"City : {city}")

        for contact in contacts:
            print(contact)

        print("-" * 50)
```

```python
def view_by_state():

    state_contacts = book.view_by_state()

    if not state_contacts:
        print("\nNo Contacts Found.")
        return

    print("\nContacts Grouped By State\n")

    for state, contacts in state_contacts.items():

        print(f"State : {state}")

        for contact in contacts:
            print(contact)

        print("-" * 50)
```

```python
def count_by_city():

    city_count = book.count_by_city()

    if not city_count:
        print("\nNo Contacts Found.")
        return

    print("\nContact Count By City\n")

    for city, count in city_count.items():
        print(f"{city} : {count}")
```

```python
def count_by_state():

    state_count = book.count_by_state()

    if not state_count:
        print("\nNo Contacts Found.")
        return

    print("\nContact Count By State\n")

    for state, count in state_count.items():
        print(f"{state} : {count}")
```
